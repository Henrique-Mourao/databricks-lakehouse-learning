<b> Monthly Order Summary
1. total orders
2. total items bought
3. total amount spent

```python
df_orders = spark.table('gizmobox.silver.py_orders')
display(df_orders)
```

```python
df_orders.printSchema()
```

```python
from pyspark.sql import functions as F

df_order_summary = (
    df_orders
    .select(
        F.col('json_value.customer_id').alias('customer_id'),
        F.col('json_value.order_id').alias('order_id'),
        F.col('json_value.transaction_timestamp').alias('transaction_timestamp'),
        F.explode('json_value.items').alias('item')
    )
    .withColumn("order_month", F.date_format('transaction_timestamp', 'yyyy-MM'))
    .groupBy('order_month', 'customer_id')
    .agg(
        F.countDistinct('order_id').alias('total_orders'),
        F.sum('item.quantity').alias('total_items_bought'),
        F.sum(F.col('item.price') * F.col('item.quantity')).alias('total_amount')
    )
)

display(from pyspark.sql import functions as F

df_order_summary = (
    df_orders
    .select(
        F.col('json_value.customer_id').alias('customer_id'),
        F.col('json_value.order_id').alias('order_id'),
        F.col('json_value.transaction_timestamp').alias('transaction_timestamp'),
        F.explode('json_value.items').alias('item')
    )
    .withColumn("order_month", F.date_format('transaction_timestamp', 'yyyy-MM'))
    .groupBy('order_month', 'customer_id')
    .agg(
        F.countDistinct('order_id').alias('total_orders'),
        F.sum('item.quantity').alias('total_items_bought'),
        F.sum(F.col('item.price') * F.col('item.quantity')).alias('total_amount')
    )
)

display(df_order_summary))
```
