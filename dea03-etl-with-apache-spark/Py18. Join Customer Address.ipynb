<b> Join Customer and Address 
<br>

Join customer data with address data to create a customer_address table whitch contains the address of each customer on the same

```python
df_customers = spark.table('gizmobox.silver.py_customers')
display(df_customers)
```

```python
df_addresses = spark.table('gizmobox.silver.py_addresses')
display(df_addresses)
```

```python
df_customers_address = (
    df_customers.join(df_addresses, "customer_id", "inner")
    .select(
        "customer_id",
        "customer_name",
        "email",
        "date_of_birth",
        "member_since",
        "telephone",
        "shipping_address_line_1",
        "shipping_city",
        "shipping_state",
        "shipping_postcode",
        "billing_address_line_1",
        "billing_city",
        "billing_state",
        "billing_postcode"
    )
)
display(df_customers_address)
```

```python
df_customers_address.writeTo("gizmobox.gold.py_customers_address").createOrReplace()
```
