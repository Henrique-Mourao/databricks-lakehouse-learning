<b>1. Read Returns Data via JDBC

```python
from pyspark.sql.types import StructType, StructField, StringType, TimestampType, DoubleType

schema = StructType([
    StructField("refund_id", StringType(), True),
    StructField("payment_id", StringType(), True),
    StructField("refund_date", TimestampType(), True),
    StructField("refund_amount", DoubleType(), True),
    StructField("refund_reason", StringType(), True),
    StructField("refund_source", StringType(), True)
])

empty_df = spark.createDataFrame([], schema)

empty_df.writeTo("gizmobox.bronze.py_refunds").createOrReplace()
```
