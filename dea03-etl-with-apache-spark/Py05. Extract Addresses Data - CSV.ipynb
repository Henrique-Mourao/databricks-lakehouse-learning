<b>Extract Data From the Address Files 
1. Read all the addresses files
2. Create addresses table in the bronze layer

<b> 1. Read all the address files

```python
df = (
      spark.read.format('csv')
          .option("header", "true")
          .option("delimiter", "\t")
          .load('/Volumes/gizmobox/landing/operational_data/addresses/')
)
display(df)
```

<b> 2. Create Addresses table in Bronze Layer

```python
df.writeTo('gizmobox.bronze.py_addresses').createOrReplace()
```
