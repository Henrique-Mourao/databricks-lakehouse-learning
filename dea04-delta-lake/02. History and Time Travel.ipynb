<b> History and Time Travel
1. Query Delta Lake table history
2. Query previous versions of the data 
3. Query data from a specific time.


<b> 1. Query Delta Lake Table History

```python
%sql 
DESCRIBE HISTORY demo.delta_lake.companies;
```

<b> 2. Query Data from a Specific Version 

```python
%sql
SELECT * FROM demo.delta_lake.companies
```

```python
%sql 
SELECT * FROM demo.delta_lake.companies
VERSION AS OF 1;   
```

<b> 3. Query Data from a Specific Time 

```python
%sql
SELECT * FROM demo.delta_lake.companies
TIMESTAMP AS OF '2022-01-01T11:00:00.000+0000'
```

<b> 4. Restore Data in the Table to a Specific Version


```python
%sql
RESTORE TABLE demo.delta_lake.companies
 VERSION AS OF 1;
```
