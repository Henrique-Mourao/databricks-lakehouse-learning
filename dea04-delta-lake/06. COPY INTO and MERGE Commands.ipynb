# <b> COPY INTO and MERGE COMMANDS
> - Incrementally loads data into Delta Lake tables from Cloud Storage 
> - Supports wide rage of file formats(CSV, JSON, Parquet, Delta)
> - Supports schema evolution 
> - Alternative to Auto Loader for batch ingestion 

Documentation - COPY INTO 



### <b> Create the table to copy the data into 

```python
%sql
CREATE TABLE IF NOT EXISTS demo.delta_lake.raw_stock_prices;
```

### <b>Incrementally load new files into the table 

```python
%sql
COPY INTO demo.delta_lake.raw_stock_prices
FROM 'abfss://demo@deacourse.dfs.core.windows.net/stock_prices'
FILEFORMAT = CSV
FORMAT_OPTIONS ('inferSchema' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true')
;
```

### <b> MERGE Statement
> - Used for upserts (Insert/ Update/ Delete operations in a single statement)
> - Allows merging new data into a target table based on matching condition

Documentation - MERGE INTO

### <b> Create the table to merge the data into 


```python
%sql
CREATE TABLE IF NOT EXISTS demo.delta_lake.stock_princes
(
    stocks_id STRING,
    price DOUBLE,
    trading_date DATE
);
```

### <b> Merge the source data into target table 
1. Insert new stocks received
2. Update prince and tranding_date if updates receieved
3. Delete stocks which are de_listed from the exchange (status='DELISTED')

```python
%sql
MERGE INTO demo.delta_lake.stock_prices AS target
USING demo.delta_lake.raw_stock_prices AS source 
    ON target.stock_id = source.stocks_id 
WHEN MATCHED AND source.status = 'ACTIVR'
```
