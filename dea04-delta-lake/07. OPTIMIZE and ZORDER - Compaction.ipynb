<b> Compaction using OPTIMIZE and Z-Ordering
> Merges multiple small files into fewer, larger files, thus improves the perfomance

<b> 1. Create table - demo.delta_lake.optimize_stock_prices 

```python
%sql
DROP TABLE IF EXISTS demo.delta_lake.optimize_stock_prices;
CREATE TABLE IF NOT EXISTS demo.delta_lake.optimize_stock_prices
(
    stock_id STRING,
    price DOUBLE, 
    trading_date DATE
);
```

<b> 2. Inserts some data in a few transactions

```python
%sql
INSERT INTO demo.delta_lake.optimize_stock_prices
VALUES ('AAPL', 225.40, '2025-02-10');
```

```python
%sql
INSERT INTO demo.delta_lake.optimize_stock_prices
VALUES ('GOOGL', 1312.30, '2025-02-10');
```

```python
%sql
INSERT INTO demo.delta_lake.optimize_stock_prices
VALUES ('MSFT', 390.00, '2025-02-10');
```

<b> 5. Run OPTIMIZE

```python
%sql 
OPTIMIZE demo.delta_lake.optimize_stock_prices;
```

```python
%sql 
OPTIMIZE demo.delta_lake.optimize_stock_prices;
ZORDER BY stock_id; 
```
