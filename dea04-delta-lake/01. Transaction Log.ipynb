Undestand the cloud storage directoru structure behind delta lake tables

<b> 0. Create a new schema under the demo catalog for this section of the course (delta_lake)


```python
%sql
CREATE SCHEMA IF NOT EXISTS demo.delta_lake
        MANAGED LOCATION 'adfss://demo@course.dfs.core.windows.net/delta_lake';
```

<b>1. Create a Delta Lake Table

```python
%sql
CREATE TABLE IF NOT EXISTS demo.delta_lake.companies
    ( 
        company_name STRING,
        founded_date DATE,
        country STRING
    );
```

```python
%sql
DESC EXTENDED demo.delta_lake.companies;
```
